```python
category_median= df.groupby('Category')['Review Rating'].median()

df['median fill'] = df['Category'].map(category_median)

df['Review Rating'] = df['Review Rating'].fillna(df['median fill'])
df =df.drop(columns='median fill')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df =df.rename(columns={'purchase_amount_(usd)':'purchase_amount_usd'})
```

```python
df.columns
```

```python
# Age bracket colmns
labels= ['Young','Young Adult','Middle-Age','Senior']

df['age_bracket']= pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_bracket']].head(50
                              )
```

```python
df['age_bracket'] ,bins = pd.qcut(df['age'], q=4 ,labels=labels , retbins=
                                  True)
print(bins)
```

```python
# create frequency patterns

frequency_map={
    'Fortnightly':14,
    'Quarterly':90,
    'Weekly': 7,
    'Annually':365,
    'Bi-Weekly':14,
    'Monthly':30,
    'Every 3 Months':90
}

df['purchase_frequency_in_days'] = df['frequency_of_purchases'].map(frequency_map)
```
